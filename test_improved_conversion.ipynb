## Process a Video with Improved Settings

Let's process the walking.mp4 video using our improved conversion method.

```python
# Create a configuration with optimal settings
config = BvhConverterConfig()
config.target_fps = 30
config.position_smoothing_window = 7  # Increased smoothing for positions
config.rotation_smoothing_window = 7  # Increased smoothing for rotations
config.rotation_smoothing_factor = 0.6  # Higher factor = more smoothing
config.enforce_joint_limits = True  # Prevent unrealistic poses
config.use_quaternions = True  # Use quaternion-based rotation calculation

# Set file paths
input_video = "videos/walking.mp4"
output_bvh = "bvh/walking_improved.bvh"

# Process the video
skeleton = process_video(input_video, output_bvh, config)
```

## Observations and Improvements

1. **Rotation Improvements**: The improved conversion uses quaternions instead of Euler angles, which should eliminate gimbal lock issues that can cause sudden jerky movements.

2. **Smoother Motion**: Enhanced temporal smoothing reduces jitter while preserving the natural motion of the walking animation.

3. **Better Joint Orientations**: Updates to the reference vectors and joint pairs result in more natural poses, particularly in the spine, chest, and head regions.

4. **Anatomical Joint Constraints**: The joint angle limits prevent physically impossible poses, ensuring the character's movements stay within realistic bounds.

These improvements should result in a BVH animation that much more closely matches the movement seen in the original video.
